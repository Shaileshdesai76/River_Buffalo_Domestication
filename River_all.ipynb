```python
no_migration_model = momi.DemographicModel(N_e=1e5, gen_time=6, muts_per_gen=1.32e-8)
sfs = momi.Sfs.load("River_all_sfs.gz")

#### from chatgpt help

no_migration_model.set_data(sfs)

#no_migration_model.add_size_param("n_CHINA")
no_migration_model.add_size_param("n_EGYPT")
no_migration_model.add_size_param("n_INDIA")
no_migration_model.add_size_param("n_IRAN")
no_migration_model.add_size_param("n_IRAQ")
no_migration_model.add_size_param("n_ITALY")
no_migration_model.add_size_param("n_RNEPAL")
no_migration_model.add_size_param("n_WNEPAL")

#no_migration_model.add_leaf("CHINA", N="n_CHINA")
no_migration_model.add_leaf("EGYPT", N="n_EGYPT")
no_migration_model.add_leaf("INDIA", N="n_INDIA")
no_migration_model.add_leaf("IRAN", N="n_IRAN")
no_migration_model.add_leaf("IRAQ", N="n_IRAQ")
no_migration_model.add_leaf("ITALY", N="n_ITALY")
no_migration_model.add_leaf("RNEPAL", N="n_RNEPAL")
no_migration_model.add_leaf("WNEPAL", N="n_WNEPAL")

## add pulse para

no_migration_model.add_pulse_param("p_R_W", lower=0.0, upper=0.1)
#no_migration_model.add_pulse_param("p_IND_W", lower=0.0, upper=0.05)
no_migration_model.add_pulse_param("p_IND_R", lower=0.0, upper=0.1)
#no_migration_model.add_pulse_param("p_IND_IRAN", lower=0.0, upper=0.1)
#no_migration_model.add_pulse_param("p_IND_IRAQ", lower=0.0, upper=0.1)
#no_migration_model.add_pulse_param("p_IND_EGYPT", lower=0.0, upper=0.05)

no_migration_model.add_pulse_param("p_IRAN_IRAQ", lower=0.0, upper=0.1)
no_migration_model.add_pulse_param("p_EGYPT_ITALY", lower=0.0, upper=0.1)

#no_migration_model.add_time_param("t_DOM", lower=4000, upper=10000)
no_migration_model.add_time_param("t_ITALY_EGYPT", lower=50, upper=7000)
no_migration_model.add_time_param("t_IRAQ_IRAN", lower=70, upper=7000)
no_migration_model.add_time_param("t_EGYPT_IRAN", lower_constraints=["t_IRAQ_IRAN","t_ITALY_EGYPT"], lower=100, upper=8000)
no_migration_model.add_time_param("t_INDIA_IRAN", lower_constraints=["t_EGYPT_IRAN","t_IRAQ_IRAN"], lower=2000, upper=8000)
no_migration_model.add_time_param("t_RNEPAL_INDIA", lower_constraints=["t_INDIA_IRAN"], lower=2000, upper=10000)
no_migration_model.add_time_param("t_WNEPAL_RNEPAL", lower_constraints=["t_RNEPAL_INDIA"], lower=2500, upper=10000)

## add time para
no_migration_model.add_time_param("t_mig_R_W", lower=50, upper_constraints=["t_WNEPAL_RNEPAL"])
#no_migration_model.add_time_param("t_mig_IND_W", lower=50, upper_constraints=["t_RNEPAL_INDIA"])
no_migration_model.add_time_param("t_mig_IND_R", lower=50, upper_constraints=["t_RNEPAL_INDIA"])
#no_migration_model.add_time_param("t_mig_IND_IRAN",lower=50,upper_constraints=["t_INDIA_IRAN"])

#no_migration_model.add_time_param("t_mig_IND_IRAQ",lower=50, upper_constraints=["t_INDIA_IRAN"])
#no_migration_model.add_time_param("t_mig_IND_EGYPT",lower=50, upper_constraints=["t_INDIA_IRAN"])
no_migration_model.add_time_param("t_mig_IRAN_IRAQ", lower=50, upper_constraints=["t_IRAQ_IRAN"])
no_migration_model.add_time_param("t_mig_EGYPT_ITALY", lower=50, upper_constraints=["t_ITALY_EGYPT"])

## add migration
# RNEPAL → WNEPAL
no_migration_model.move_lineages("RNEPAL", "WNEPAL", t="t_mig_R_W", p="p_R_W")
# INDIA → WNEPAL (small)
#no_migration_model.move_lineages("WNEPAL", "INDIA", t="t_mig_IND_W", p="p_IND_W")
# INDIA → RNEPAL (slightly larger)
no_migration_model.move_lineages("RNEPAL", "INDIA", t="t_mig_IND_R", p="p_IND_R")
# INDIA → IRAN
#no_migration_model.move_lineages("INDIA", "IRAN", t="t_mig_IND_IRAN", p="p_IND_IRAN")

# INDIA → IRAQ
#no_migration_model.move_lineages("INDIA", "IRAQ", t="t_mig_IND_IRAQ", p="p_IND_IRAQ")

# INDIA → EGYPT
#no_migration_model.move_lineages("INDIA", "EGYPT", t="t_mig_IND_EGYPT", p="p_IND_EGYPT")

# IRAN ↔ IRAQ (choose one direction unless you want two pulses)
no_migration_model.move_lineages("IRAN", "IRAQ", t="t_mig_IRAN_IRAQ", p="p_IRAN_IRAQ")

# EGYPT ↔ ITALY
no_migration_model.move_lineages("EGYPT", "ITALY", t="t_mig_EGYPT_ITALY", p="p_EGYPT_ITALY")

#no_migration_model.add_time_param("t_CHINA_WNEPAL", lower_constraints=["t_WNEPAL_RNEPAL"])

#no_migration_model.move_lineages("WNEPAL","CHINA",t="t_CHINA_WNEPAL")
no_migration_model.move_lineages("RNEPAL","WNEPAL",t="t_WNEPAL_RNEPAL")
no_migration_model.move_lineages("INDIA","RNEPAL",t="t_RNEPAL_INDIA")
no_migration_model.move_lineages("IRAN","INDIA",t="t_INDIA_IRAN")
no_migration_model.move_lineages("EGYPT","IRAN",t="t_EGYPT_IRAN")
no_migration_model.move_lineages("IRAQ","IRAN",t="t_IRAQ_IRAN")
no_migration_model.move_lineages("ITALY","EGYPT",t="t_ITALY_EGYPT")

print("\n===== STOCHASTIC OPTIMIZATION =====")
no_migration_model.stochastic_optimize(
    snps_per_minibatch=4000,
    num_iters=100,
    svrg_epoch=5
)


print("\n===== FULL OPTIMIZATION =====")
res = no_migration_model.optimize(
    method="L-BFGS-B",
    options={"maxiter": 400, "maxfun": 1000}
)
```

```python
model = momi.DemographicModel(N_e=1e5, gen_time=6,
                              muts_per_gen=1.32e-8)

model.set_data(sfs)

model.add_size_param("n_china")
model.add_size_param("n_india")
model.add_size_param("n_wnepal")
model.add_size_param("n_rnepal")
#model.add_size_param("n_italy")
#model.add_size_param("n_egypt")
#model.add_size_param("n_iran")
#model.add_size_param("n_iraq")

#model.add_time_param("t_iraq_iran")
#model.add_time_param("t_italy_egypt")
#model.add_time_param("t_iran_egypt", lower_constraints=["t_italy_egypt"])
#model.add_time_param("t_egypt_india", lower_constraints=["t_iran_egypt"])
model.add_time_param("t_india_rnepal")
model.add_time_param("t_rnepal_wnepal", lower_constraints=["t_india_rnepal"])
model.add_time_param("t_wnepal_china", lower_constraints=["t_rnepal_wnepal"])

model.add_leaf("CHINA", N="n_china")
model.add_leaf("INDIA", N="n_india")
model.add_leaf("WNEPAL", N="n_wnepal")
model.add_leaf("RNEPAL", N="n_rnepal")



model.move_lineages("INDIA", "RNEPAL", t="t_india_rnepal")
model.move_lineages("RNEPAL", "WNEPAL", t="t_rnepal_wnepal")
model.move_lineages("WNEPAL", "CHINA", t="t_wnepal_china")

model.optimize()
```
