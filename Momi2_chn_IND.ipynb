```python
migration_model_2 = momi.DemographicModel(
    N_e=1e6,
    gen_time=6,
    muts_per_gen=1.32e-8
)

migration_model_2.set_data(sfs)

# ---- Parameters ----
migration_model_2.add_size_param("n_north")
migration_model_2.add_size_param("n_south")

migration_model_2.add_time_param("tdiv")
migration_model_2.add_time_param("tdiv_dome", upper=10000)

# migration must be more recent than divergence
migration_model_2.add_time_param("tmig_north_south", upper_constraints=["tdiv_dome"])
migration_model_2.add_time_param("tmig_south_north", upper_constraints=["tdiv_dome"])

migration_model_2.add_pulse_param("mfrac_north_south", lower=0.2)
migration_model_2.add_pulse_param("mfrac_south_north", lower=0.2)

# ---- Leaves ----
migration_model_2.add_leaf("CHINA", N="n_north")
migration_model_2.add_leaf("INDIA", N="n_south")

# ---- Migration pulses (BEFORE split) ----
migration_model_2.move_lineages(
    "CHINA", "INDIA",
    t="tmig_north_south",
    p="mfrac_north_south"
)

migration_model_2.move_lineages(
    "INDIA", "CHINA",
    t="tmig_south_north",
    p="mfrac_south_north"
)

# ---- Divergence ----
migration_model_2.move_lineages(
    "INDIA", "CHINA",
    t="tdiv"
)

# ---- Optimize ----
migration_model_2.optimize()
```

```python
migration_model = momi.DemographicModel(
    N_e=1e5,
    gen_time=6,
    muts_per_gen=2.2e-9
)

migration_model.set_data(sfs)

# Parameters
migration_model.add_size_param("n_china")
migration_model.add_size_param("n_india")
migration_model.add_size_param("n_ghost")

migration_model.add_time_param("tdiv")
migration_model.add_time_param("tmig", upper_constraints=["tdiv"])


migration_model.add_pulse_param("mfrac", upper=0.2)

# Leaves
migration_model.add_leaf("CHINA", N="n_china")
migration_model.add_leaf("INDIA", N="n_india")

# Ghost population (NO DATA)
migration_model.add_leaf("ghostCHINA", N="n_ghost")

# Migration involving ghost
migration_model.move_lineages(
    "CHINA", "ghostCHINA",
    t="tmig",
    p="mfrac"
)

# Divergence
migration_model.move_lineages(
    "INDIA", "CHINA", t="tdiv"
)
migration_model.move_lineages(
    "ghostCHINA", "CHINA", t="tdiv"
)

# ---- Optimize ----
migration_model.optimize()
```

```python
model = momi.DemographicModel(N_e=1e5, gen_time=6)

# add ANCESTRAL
model.add_leaf("CHINA", N=1e5)
# ADDING LEAF
model.add_leaf("INDIA", N=1e5)
model.add_leaf("WNEPAL", N=1e5)
model.add_leaf("RNEPAL", N=1e5)
model.add_leaf("ITALY", N=1e5)
model.add_leaf("EGYPT", N=1e5)
model.add_leaf("IRAN", N=1e5)
model.add_leaf("IRAQ", N=1e5)
# JOINING AND MERGING 
model.move_lineages("IRAQ", "IRAN", t=100)
model.move_lineages("ITALY", "EGYPT", t=300)
model.move_lineages("IRAN", "EGYPT", t=1000)
model.move_lineages("EGYPT", "INDIA", t=4000)
model.move_lineages("INDIA", "RNEPAL", t=7000)
model.move_lineages("RNEPAL", "WNEPAL", t=10000)
model.move_lineages("WNEPAL", "CHINA", t=540000)

```
